```python
#Find Duplicates in the list
numbers = [10, 20, 30, 20, 40, 10, 50]

duplicates = []

for num in numbers:
    if numbers.count(num) > 1 and num not in duplicates:
        duplicates.append(num)

print("Duplicate elements:", duplicates)
```

```python
#Find second largest number without using sort()

numbers =[10,20,160,129,56,34,97,156]

largest = max(numbers)

second_largest = None

for i in numbers:
    if i != largest:
        if second_largest is None or i > second_largest:
            second_largest = i

print("Secondlargest number:", second_largest)
```

```python
#check whether two strings are anagrams

word1 = input("enter the word:")
word2 = input("enter the word:")

word1 = word1.lower()
word2 = word2.lower()

if sorted(word1) == sorted(word2):
    print("The strings are anagrams")
else:
    print("The strings are not anagrams")
```
