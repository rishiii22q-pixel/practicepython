```python
#swap two variables

a = input("enter the first variable")
b = input("enter the second variable")
print(f"original varibale a = {a} and b = {b}")
temp = a
a = b
b = temp
print(f"The swapped variables of a = {a} and b = {b} ")
```

```python
#largest and smallest value in list

list = [45,90,12,56,9,105,78,90]

largest = list[0]     #here we initialize the largest value[0] = 45
smallest = list[0]      #same we initialize [0] =45 in the list

for list in list:     #here loop starts for list
    if list>largest:          #comparing each value in loop and stored 
        largest = list
    if list < smallest:
        smallest = list


print("largest value is", largest)
print("smallest value is",smallest)
```

```python
#removing duplicates
numbers = [20,60,10,78,20,19,76,60] #given duplicates values
removed = set(numbers)      #using set we can eliminate duplicates 
print(removed)      #here output will be in unordered because of set
```

```python
#common element between two lists
list1 = [10,20,30,40,50,60]
list2 = [30,40,50,70,80]

common = []  # here initialize the common 

for i in list1:      #loop starts 
    if i in list2:     #it will compare the both lists if true it will be append
        common.append(i)  #here append values are stored in common 

print("common values in list are", common)
```

```python
#who is the highest mark using dictionary

students ={"Rishi":67, "Abi":78, "Priya":80,"Praveen":85,"Rahul":65}

highest = max(students,key=students.get)
print("Highest:", highest)
print("Marks:",students[highest])
```
